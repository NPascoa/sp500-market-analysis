# 3 - CLEANING

```python
import pandas as pd
from pathlib import Path 

source = Path(r"filename\csv_basic_clean")

companies=pd.read_csv(source / "companies_bclean.csv")
index=pd.read_csv(source / "index_bclean.csv")
stocks=pd.read_csv(source / "stocks_bclean.csv")
```

After the findings from the exploration phase, this is the course of action for cleaning phase:
1. Exclude all rows where all value columns (adj_close, high, low, open, close and volume) were null (5.3% of stocks rows)
2. Retain the volume=0 and flat value columns rows, but flag them.

Nothing will be done to Companies and Index tables.

```python
stocks.columns
```

```python
#STOCKS CLEAN TABLE:

raw_row_count=len(stocks)
#1891536

#removing full null rows for the specified columns:
stocks_clean = stocks.dropna(subset=["open", "high", "low", "close", "adj_close","volume"], how="all").copy()

stocks_clean
```

```python
#add flag column:
stocks_clean["no_trading"]=(
    (stocks_clean["volume"]==0) &
    (stocks_clean["open"]==stocks_clean["high"])&
    (stocks_clean["high"]==stocks_clean["low"])&
    (stocks_clean["low"]==stocks_clean["close"])
)

stocks_clean[stocks_clean["no_trading"]]
```

Verify:

```python
print("OG Stocks rows:",len(stocks))
print("Clean Stocks rows:",len(stocks_clean))
print("Result confirm:",len(stocks) - stocks[["open", "high", "low", "close", "adj_close","volume"]].isna().all(axis=1).sum())
print("flagged as no_trading:", stocks_clean["no_trading"].sum())
```

Saving cleaned tables:

```python
destination = Path(r"filename\csv_clean")

companies.to_csv(destination / "companies_clean.csv", index=False)
index.to_csv(destination / "index_clean.csv", index=False)
stocks_clean.to_csv(destination / "stocks_clean.csv", index=False)
```
