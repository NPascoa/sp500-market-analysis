## LOADING:

```python
import pandas as pd
```

Loading CSVs:

```python
companies = pd.read_csv(r"filepath\csv_raw\sp500_companies.csv")
index = pd.read_csv(r"filepath\csv_raw\sp500_index.csv")
stocks = pd.read_csv(r"filepath\csv_raw\sp500_stocks.csv")

print("tables_size:\n","companies:",companies.shape,"\n","index:",index.shape,"\n", "stocks:",stocks.shape)
#could use len(companies) to get just the number of rows

print(companies.head())
print(index.head())
print(stocks.head())

```

The len() of the tables match the counts in sql.

## CLEANING NAMES AND DATA TYPES

Companies table:

```python
#all to lower case - it will be easier to query:
companies.columns = companies.columns.str.lower()
#add _ to word joins:
companies = companies.rename(columns={
    "shortname":"short_name",
    "longname":"long_name",
    "currentprice":"current_price",
    "market_cap":"market_cap",
    "revenuegrowth":"revenue_growth",
    "fulltimeemployees":"fulltime_employees",
    "longbusinesssummary":"long_business_summary"
})

companies
```

Index table:

```python
#renaming the columns:
index = index.rename(columns={
    "Date":"date",
    "S&P500":"sp500"
})

#fixing the date type:
index["date"]=pd.to_datetime(index["date"])

index
```

Stocks table:

```python
#lowering columns case:
stocks.columns = stocks.columns.str.lower()

stocks = stocks.rename(columns={
    "adj close":"adj_close"
})

stocks["date"]=pd.to_datetime(stocks["date"])

stocks
```

Save the csv files after the basic cleaning:

```python
from pathlib import Path #to create a path

destination = Path(r"filepath\csv_basic_clean")


companies.to_csv(destination / "companies_bclean.csv", index=False)
index.to_csv(destination / "index_bclean.csv", index=False)
stocks.to_csv(destination / "stocks_bclean.csv", index=False)
```
